```python
import pandas as pd
```

```python
taxonomy = pd.read_csv(
    "dada2rb-taxonomy-annotate.txt",
    sep="\t",
    header=None,
    names=["id", "taxonomy"]
)

counts = pd.read_csv(
    "dada2rb-genus-count.txt",
    sep="\t",
    index_col=0
)
```

```python
taxonomy = taxonomy[taxonomy["id"].str.contains("genus")]
```

```python
taxonomy["genus"] = taxonomy["taxonomy"].str.split(";").str[-1]
taxonomy.head()
```

```python
mapping = dict(
    zip(
        taxonomy["id"],
        taxonomy["genus"]
    )
)
```

```python
counts = counts.rename(columns=mapping)
```

```python
counts.columns = [
    "g__" + str(x)
    for x in counts.columns
]
```

```python
counts = counts.drop(columns=counts.filter(like="g__NA").columns)
```

```python
counts
```

```python
counts = counts.T
counts.head()
```

```python
counts.to_csv(
    "NHANES_MGM_genus_counts.csv",
    index=True
)
```

```python
tax_dict_azh = pd.read_csv(
    "taxonomy.tsv",
    sep="\t"
)
tax_dict_azh.head()
```

```python
def extract_genus(tax):
    levels = str(tax).split(";")

    for level in levels:
        level = level.strip()
        if level.startswith("g__"):
            return level

    return None

tax_dict_azh["genus"] = tax_dict_azh["Taxon"].apply(extract_genus)
```

```python
df_asv = pd.read_csv('asv_table.tsv', sep = '\t', header=1)
runs = df_asv.columns
df_asv.head()
```

```python
mapping_azh = dict(
    zip(
        tax_dict_azh["Feature ID"],
        tax_dict_azh["genus"]
    )
)
new_cols = df_asv.iloc[:, 0]
df_asv = df_asv.iloc[:, 1:].T
df_asv.columns = new_cols.values
df_asv = df_asv.rename(columns=mapping_azh)

df_asv = df_asv.groupby(df_asv.columns, axis=1).sum()


df_asv.head()
```

```python
manual_mapping = {
    'g__Cutibacterium': 'g__Propionibacterium',
    'g__Burkholderia-Caballeronia-Paraburkholderia': 'g__Burkholderia',


    "g__Pseudopropionibacterium": "g__Propionibacterium",
    "g__Eubacterium_saphenum_group": "g__Eubacterium",
    "g__Eubacterium_yurii_group": "g__Eubacterium",

    "g__Saccharimonadaceae": "g__Candidatus_Saccharimonas",
    "g__Saccharimonadales": "g__Candidatus_Saccharimonas",
    "g__TM7a": "g__Candidatus_Saccharimonas",
    "g__TM7x": "g__Candidatus_Saccharimonas",

    "g__Prevotellaceae": "g__Prevotella",
    "g__Prevotellaceae_UCG_004": "g__Prevotella",

    "g__Escherichia": "g__Escherichia_Shigella",
    "g__Shigella": "g__Escherichia_Shigella",

    "g__S24_7": "g__Muribaculaceae",
    "g__S24-7": "g__Muribaculaceae",

    "g__Bacilli_RF39": "g__RF39",
    "g__Mollicutes_RF39": "g__RF39",
    "g__Izemoplasmatales": "g__RF39",
}


df_asv.columns = (
    df_asv.columns
    .map(lambda x: manual_mapping.get(x, x))  # Aplicar el diccionario
    .str.replace("-", "_", regex=False)       # Sustituir - por _
    .str.replace("[", "", regex=False)        # Eliminar [
    .str.replace("]", "", regex=False)        # Eliminar ]
)


df_asv = df_asv.groupby(df_asv.columns, axis=1).sum()

df_asv.head()
```

```python
df_asv = df_asv.T
df_asv.head()
```

```python
df_asv.to_csv("asv_genus.csv", index=True)
```

```python
df_metadata = pd.read_csv('SraRunTable.csv')

df_metadata["disease_status"] = (
    df_metadata["Sample Name"]
    .str.extract(r"(Patient|Control)", expand=False)
    .str.lower()
)

df_metadata["disease_status"] = df_metadata["disease_status"].map({
    "control": 0,
    "patient": 1
})

df_metadata["pair_id"] = (
    df_metadata["Sample Name"]
    .str.extract(r"(\d+)", expand=False)
    .astype(int)
)

df_metadata = df_metadata[["Run","disease_status","pair_id"]]

df_metadata
```

```python
df_metadata = df_metadata[df_metadata['Run'].isin(runs)]
```

```python
df_metadata.to_csv("df_metadata.csv", index=True)
```
